```python
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd
```

```python
data_gardener_A = {
    'Fertilizer A': [15.2, 17.3, 16.5, 16.8, 17.1, 15.6],
    'Fertilizer B': [18.4, 17.8, 19.1, 18.2, 19.4, 18.5],
    'Fertilizer C': [20.1, 21.3, 19.9, 20.5, 21.1, 20.8]
}

data_gardener_B = {
    'Fertilizer A': [14.5, 15.0, 16.2, 15.1, 15.9, 14.7],
    'Fertilizer B': [18.0, 18.3, 17.7, 18.6, 19.0, 18.2],
    'Fertilizer C': [19.7, 20.3, 20.5, 20.1, 21.2, 20.4]
}
```

```python
df_gardener_A=pd.DataFrame(data_gardener_A)
df_gardener_B=pd.DataFrame(data_gardener_B)
```
